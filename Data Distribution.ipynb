## 1. Normal Distribution

### Definition

A normal distribution is a continuous probability distribution that is symmetric around its mean.

It has a characteristic bell-shaped curve.

In a perfectly normal distribution:

- Mean = Median = Mode
- The distribution is symmetric.
- Most observations are concentrated around the mean.
- Fewer observations occur as we move away from the mean.

### Formula

The probability density function of a normal distribution is:

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $\mu$ = mean
- $\sigma$ = standard deviation
- $x$ = observation

### Numerical Example

Suppose exam scores have:

Mean = 70

Standard deviation = 10

A score of 70 is exactly at the mean.

A score of 80 is one standard deviation above the mean.

A score of 60 is one standard deviation below the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=70, scale=10, size=1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

A company may analyze employee performance scores. If the scores approximately follow a normal distribution, most employees will have scores close to the average, while relatively fewer employees will have extremely high or low scores.

### AI/ML Example

Normal distributions are important in statistical modeling and machine learning. They are also commonly encountered when analyzing numerical features and model errors.

For example, residuals from a regression model are often examined to determine whether they approximately follow a normal distribution.

### Interpretation

A normal distribution is centered around its mean and has a symmetric bell-shaped pattern. Values close to the mean occur more frequently than values far from the mean.

## 2. Uniform Distribution

### Definition

A uniform distribution is a probability distribution in which values within a specified interval have equal probability density.

There are two types:

- Discrete Uniform Distribution
- Continuous Uniform Distribution

### Formula

For a continuous uniform distribution between $a$ and $b$:

$$
f(x)=\frac{1}{b-a}
$$

where:

$$
a \leq x \leq b
$$

### Numerical Example

Suppose a random number is selected between 1 and 10 from a continuous uniform distribution.

Every value within the interval has the same probability density.

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

A business simulation may use a uniform distribution when every value within a known range is considered equally likely.

For example, a simple simulation may assume that customer arrival time within a particular short interval is equally likely.

### AI/ML Example

Uniform distributions can be used to generate random values during simulations and to initialize parameters in some machine learning algorithms.

### Interpretation

A uniform distribution has approximately equal probability across its defined range, unlike a normal distribution where values near the mean are more common.

## 3. Gaussian Distribution

### Definition

The Gaussian distribution is another name for the normal distribution.

It is a continuous, symmetric, bell-shaped probability distribution characterized by its mean and standard deviation.

### Important Characteristics

- Mean determines the center.
- Standard deviation determines the spread.
- It is symmetric around the mean.
- Mean = Median = Mode.

### Numerical Example

Suppose:

Mean = 100

Standard deviation = 15

Then:

- 100 is the center.
- 85 is one standard deviation below the mean.
- 115 is one standard deviation above the mean.

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(100, 15, 1000)

plt.hist(data, bins=30)
plt.title("Gaussian Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

A company can use a Gaussian model as an approximation when analyzing a numerical measurement that is approximately bell-shaped, such as certain quality-control measurements.

### AI/ML Example

Gaussian distributions are used in many statistical and machine learning methods, including Gaussian Naive Bayes and Gaussian mixture models.

### Interpretation

Gaussian distribution and normal distribution describe the same family of bell-shaped distributions. The mean controls the center and the standard deviation controls the spread.

## 4. Left Skew

### Definition

A left-skewed distribution, also called a negatively skewed distribution, has a longer tail on the left side.

Most observations are concentrated toward the higher-value side.

### Relationship

In a typical left-skewed distribution:

$$
Mean < Median < Mode
$$

### Numerical Example

Consider a dataset where most students score highly, but a small number of students have very low scores.

The low scores create a longer tail toward the left.

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = -np.random.exponential(scale=1, size=1000)

plt.hist(data, bins=30)
plt.title("Left-Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

If most customers spend relatively high amounts but a small number of customers spend very little, the distribution may be left-skewed.

### AI/ML Example

During exploratory data analysis, identifying skewness helps data scientists understand feature distributions and decide whether transformations may be useful.

### Interpretation

A left-skewed distribution has a longer tail toward lower values. The mean is generally pulled toward the left tail.

## 5. Right Skew

### Definition

A right-skewed distribution, also called a positively skewed distribution, has a longer tail on the right side.

Most observations are concentrated toward the lower-value side.

### Relationship

In a typical right-skewed distribution:

$$
Mode < Median < Mean
$$

### Numerical Example

Income data is often used as an example where most people have lower or moderate incomes while a smaller number have very high incomes.

These high values create a long right tail.

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=1, size=1000)

plt.hist(data, bins=30)
plt.title("Right-Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

Customer spending can be right-skewed when most customers spend moderate amounts but a small number of customers make very large purchases.

### AI/ML Example

Right-skewed numerical features may be transformed using techniques such as logarithmic transformation when appropriate, especially when the feature has a long positive tail.

### Interpretation

A right-skewed distribution has a longer tail toward higher values. Extreme high values tend to pull the mean toward the right.

## 6. Symmetric Distribution

### Definition

A distribution is symmetric when its left and right sides have approximately the same shape around the center.

For a perfectly symmetric unimodal distribution:

$$
Mean = Median = Mode
$$

### Numerical Example

Consider:

**10, 20, 30, 40, 50**

The distribution is centered around 30.

### Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.title("Symmetric Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

### Business Example

A business may observe a relatively symmetric distribution in a measurement where values are balanced around a central value.

### AI/ML Example

Understanding whether a feature is symmetric or skewed helps during exploratory data analysis and can guide appropriate preprocessing choices.

### Interpretation

A symmetric distribution has similar patterns on both sides of its center. A normal distribution is an important example of a symmetric distribution.aa

## 7. Kurtosis

### Definition

Kurtosis is a statistical measure that describes aspects of the shape of a distribution, particularly the heaviness of its tails relative to a normal distribution.

It can help identify whether a dataset has relatively light or heavy tails.

### Types of Kurtosis

1. Mesokurtic
2. Leptokurtic
3. Platykurtic

### Mesokurtic

A normal distribution is commonly used as the reference for mesokurtic behavior.

### Leptokurtic

A leptokurtic distribution has relatively heavy tails and can have more extreme observations than a normal distribution.

### Platykurtic

A platykurtic distribution has relatively light tails compared with a normal distribution.

### Python Implementation

```python
import numpy as np
from scipy.stats import kurtosis

data = np.random.normal(0, 1, 1000)

kurtosis_value = kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

### AI/ML Example

Kurtosis can be used during exploratory data analysis to understand the shape and tail behavior of numerical features and identify features that may contain extreme observations.

### Business Example

In finance, kurtosis can be examined when studying investment returns because heavy tails may indicate a greater presence of extreme returns than expected under a normal model.

### Interpretation

Kurtosis helps describe the tail behavior of a distribution. High positive excess kurtosis indicates relatively heavy tails, while negative excess kurtosis indicates relatively light tails.

## 8. Empirical Rule (68–95–99.7)

### Definition

The empirical rule describes how data is distributed around the mean in an approximately normal distribution.

According to the rule:

- Approximately 68% of observations lie within 1 standard deviation of the mean.
- Approximately 95% lie within 2 standard deviations.
- Approximately 99.7% lie within 3 standard deviations.

### Formula

For mean $\mu$ and standard deviation $\sigma$:

### Within 1 Standard Deviation

$$
\mu-\sigma \leq X \leq \mu+\sigma
$$

Approximately **68%** of observations.

### Within 2 Standard Deviations

$$
\mu-2\sigma \leq X \leq \mu+2\sigma
$$

Approximately **95%** of observations.

### Within 3 Standard Deviations

$$
\mu-3\sigma \leq X \leq \mu+3\sigma
$$

Approximately **99.7%** of observations.

### Numerical Example

Suppose exam scores are approximately normally distributed with:

Mean = 70

Standard deviation = 10

Approximately:

68% of scores are between:

$$
70-10 = 60
$$

and

$$
70+10 = 80
$$

So approximately 68% of scores lie between **60 and 80**.

Approximately 95% lie between:

**50 and 90**

Approximately 99.7% lie between:

**40 and 100**

```python
mean = 70
std = 10

print("68% range:", mean - std, "to", mean + std)
print("95% range:", mean - 2 * std, "to", mean + 2 * std)
print("99.7% range:", mean - 3 * std, "to", mean + 3 * std)
```

### Business Example

A quality-control team can use the empirical rule to understand how measurements are distributed around the average when the process is approximately normal.

Values far from the mean may be investigated as unusual observations.

### AI/ML Example

The empirical rule can help data scientists understand normally distributed features and identify observations that are unusually far from the mean.

For example, observations beyond approximately three standard deviations can be investigated as potential anomalies, although this should not automatically mean that they are errors.

### Interpretation

The 68–95–99.7 rule provides a quick way to understand the spread of approximately normally distributed data.

It should be applied when the distribution is reasonably close to normal.
