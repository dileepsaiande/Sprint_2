## 1. What is Statistics?

Statistics is the branch of mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data.

It helps us understand patterns and relationships in data and supports decision-making based on evidence.

### Example

Suppose a company collects the salaries of 100 employees. Statistics can be used to calculate the average salary, identify the highest and lowest salaries, and understand how salaries are distributed.

## 2. Why Statistics is Important in AI/ML

Statistics is important in AI and Machine Learning because ML models learn patterns from data.

Statistics helps in understanding datasets, identifying patterns, detecting outliers, handling uncertainty, selecting useful features, and evaluating model performance.

### Example

Before training a machine learning model, we can use statistical measures such as mean, median, standard deviation, and correlation to understand the dataset.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
```

## 3. Statistics vs Machine Learning

Statistics and Machine Learning both work with data, but their main objectives are different.

| Statistics | Machine Learning |
|---|---|
| Focuses on understanding and interpreting data | Focuses on learning patterns from data |
| Often used for inference and hypothesis testing | Often used for prediction and automation |
| Uses statistical models | Uses algorithms and models |
| Emphasizes interpretation and uncertainty | Emphasizes predictive performance |

### Example

Statistics can help determine whether two variables are significantly related, while Machine Learning can use those variables to predict an outcome.

## 4. Descriptive Statistics

Descriptive statistics summarizes and describes the important characteristics of a dataset.

Common descriptive statistical measures include:

- Mean
- Median
- Mode
- Range
- Variance
- Standard Deviation
- Minimum and Maximum

### Example

For the data [10, 20, 30, 40, 50], the mean is 30 and the median is 30.

```python
import statistics

data = [10, 20, 30, 40, 50]

print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode([10, 20, 20, 30, 40]))
```

## 5. Inferential Statistics

Inferential statistics is used to draw conclusions or make predictions about a larger population using information collected from a sample.

It includes methods such as:

- Hypothesis testing
- Confidence intervals
- Regression analysis
- Statistical significance testing

### Example

A company surveys 500 customers to understand the satisfaction level of all its customers. The sample results can be used to make conclusions about the larger customer population.

## 6. Population

A population is the complete group of individuals, objects, or observations that we want to study.

### Example

If we want to study the salaries of all employees in a company, all employees in that company represent the population.

## 7. Sample

A sample is a smaller subset selected from a population for the purpose of analysis.

Studying a sample is often faster and less expensive than studying the entire population.

### Example

If a company has 10,000 employees and we collect salary information from 500 employees, those 500 employees represent the sample.

## 8. Census

A census is the process of collecting information from every member of a population.

### Example

If a company collects salary information from all 10,000 employees, it is conducting a census.

## 9. Sampling

Sampling is the process of selecting a subset of observations from a larger population.

Sampling allows researchers and data scientists to analyze a manageable amount of data while attempting to represent the larger population.

### Common Sampling Methods

- Simple Random Sampling
- Systematic Sampling
- Stratified Sampling
- Cluster Sampling

## 10. Variables

A variable is a characteristic or attribute that can have different values for different observations.

### Examples

- Age
- Salary
- Gender
- City
- Number of purchases

For example, if age is being studied, different people can have different age values.

## 11. Features

Features are input variables used by a Machine Learning model to make predictions.

### Example

For a house price prediction model, possible features include:

- Number of bedrooms
- Number of bathrooms
- Area in square feet
- Location
- Year built

The model uses these features to predict the house price.

## 12. Observation

An observation is a single recorded instance or data point in a dataset.

### Example

Consider a dataset containing information about houses. Each row representing one house is an observation.

If a dataset contains information about 1,000 houses, it contains 1,000 observations.

# Practical Example

Let's create a small dataset and use Python to calculate basic descriptive statistics.

```python
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)

df
```

```python
print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Minimum Marks:", df["Marks"].min())
print("Maximum Marks:", df["Marks"].max())
print("Standard Deviation:", df["Marks"].std())
```
