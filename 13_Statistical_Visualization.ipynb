# Notebook 13 – Statistical Data Visualization

## Introduction

Statistical data visualization helps represent data graphically to identify patterns, trends, distributions, relationships, and outliers.

### Topics Covered

1. Histogram
2. Bar Chart
3. Line Chart
4. Scatter Plot
5. Pie Chart
6. Density Plot
7. Box Plot
8. Violin Plot
9. Pair Plot
10. Heatmap

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample dataset
data = pd.DataFrame({
    "Age":[22,25,28,30,35,40,45,50],
    "Salary":[30000,35000,40000,45000,55000,65000,75000,85000],
    "Department":["IT","HR","IT","Sales","HR","IT","Sales","IT"],
    "Experience":[1,2,3,4,6,8,10,12]
})

data
```

## 1. Histogram

### Purpose
Shows the distribution of numerical data.

### When to Use
Use it to understand frequency and data spread.

### Advantages
- Easy to identify distribution
- Detect skewness

### Limitations
- Works only for numerical data

### Interpretation
The tallest bars represent the most frequent values.

```python
plt.hist(data["Salary"], bins=5)
plt.title("Histogram of Salary")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

## 2. Bar Chart

### Purpose
Compares categorical values.

### When to Use
Compare departments or categories.

### Advantages
Easy comparison.

### Limitations
Not suitable for continuous distributions.

### Interpretation
Higher bars indicate larger values.

```python
dept = data["Department"].value_counts()

plt.bar(dept.index, dept.values)
plt.title("Employees by Department")
plt.show()
```

## 3. Line Chart

### Purpose
Shows trends over time.

### When to Use
Sales, revenue, or stock prices.

### Advantages
Easy trend visualization.

### Limitations
Requires ordered data.

### Interpretation
The line shows increasing or decreasing trends.

```python
plt.plot(data["Experience"], data["Salary"], marker="o")
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

## 4. Scatter Plot

### Purpose
Shows the relationship between two numerical variables.

### When to Use
Check correlation.

### Advantages
Detects clusters and outliers.

### Limitations
Works only with numerical data.

### Interpretation
An upward pattern indicates positive correlation.

```python
plt.scatter(data["Age"], data["Salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

## 5. Pie Chart

### Purpose
Displays percentage contribution of categories.

### When to Use
Market share or department distribution.

### Advantages
Easy percentage visualization.

### Limitations
Not suitable for many categories.

### Interpretation
Each slice represents a proportion of the whole.

```python
plt.pie(dept.values, labels=dept.index, autopct="%1.1f%%")
plt.title("Department Distribution")
plt.show()
```

## 6. Density Plot

### Purpose
Shows the probability density of numerical data.

### When to Use
Understand smooth data distribution.

### Advantages
Better than histogram for continuous distribution.

### Limitations
Sensitive to small datasets.

### Interpretation
Peaks represent areas of higher data concentration.

```python
sns.kdeplot(data["Salary"], fill=True)
plt.title("Density Plot of Salary")
plt.show()
```

## 7. Box Plot

### Purpose
Detects outliers and spread.

### When to Use
Compare distributions.

### Advantages
Shows median, quartiles, and outliers.

### Limitations
Doesn't show exact frequencies.

### Interpretation
Points outside whiskers are potential outliers.

```python
sns.boxplot(y=data["Salary"])
plt.title("Box Plot")
plt.show()
```

## 8. Violin Plot

### Purpose
Combines box plot and density plot.

### When to Use
Compare distributions across categories.

### Advantages
Shows distribution shape.

### Limitations
Slightly harder to interpret.

### Interpretation
Wider areas indicate higher density.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Correlation matrix
corr = data.corr(numeric_only=True)

# Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(
    corr,
    annot=True,
    cmap="YlGnBu",      # Blue-Green color
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

## 9. Pair Plot

### Purpose
Shows relationships between multiple numerical features.

### When to Use
Exploratory Data Analysis (EDA).

### Advantages
Visualizes multiple correlations simultaneously.

### Limitations
Becomes crowded with many features.

### Interpretation
Diagonal shows distributions; off-diagonal shows relationships.

```python
sns.pairplot(data)
plt.show()
```

## 10. Heatmap

### Purpose
Visualizes the correlation matrix.

### When to Use
Find highly correlated features.

### Advantages
Easy correlation visualization.

### Limitations
Works only with numerical features.

### Interpretation
Darker colors represent stronger correlations.

```python
corr = data.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
