# Notebook 2 – Measures of Central Tendency

## Introduction

Measures of Central Tendency are statistical measures used to represent the central or typical value of a dataset.

The main measures covered in this notebook are:

1. Mean
2. Median
3. Mode
4. Weighted Mean
5. Geometric Mean
6. Harmonic Mean

Each concept is explained using definitions, formulas, numerical examples, Python implementations, business examples, AI/ML applications, and interpretations.

## 1. Mean

### Definition

The mean, commonly called the average, is calculated by adding all the values in a dataset and dividing the total by the number of values.

The mean is one of the most commonly used measures of central tendency.

### Formula

$$
Mean = \frac{\sum_{i=1}^{n} x_i}{n}
$$

Where:

- $x_i$ = individual values
- $n$ = number of observations

### Numerical Example

Consider the following values:

**10, 20, 30, 40, 50**

Mean:

$$
Mean = \frac{10+20+30+40+50}{5}
$$

$$
Mean = \frac{150}{5} = 30
$$

Therefore, the mean is **30**.

```python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

print("Mean:", mean)
```

### Business Example

A company records the monthly sales revenue for five months:

₹100,000, ₹120,000, ₹110,000, ₹130,000, and ₹140,000.

The mean sales revenue helps management understand the average monthly sales during this period.

### AI/ML Example

In a machine learning dataset, the mean can be used to understand the average value of a numerical feature.

For example, if a dataset contains the ages of customers, the mean age can provide a quick summary of the typical customer age.

The mean is also commonly used in preprocessing techniques such as mean imputation for missing numerical values.

### Interpretation

The mean represents the average value of the dataset. In the numerical example, the average value is 30.

However, the mean can be strongly affected by extreme values or outliers. Therefore, for highly skewed data, the median may sometimes provide a more representative measure of the center.

## 2. Median

### Definition

The median is the middle value of a dataset after the values have been arranged in ascending or descending order.

If the number of observations is even, the median is calculated by taking the average of the two middle values.

### Formula

For an odd number of observations:

$$
Median = x_{\frac{n+1}{2}}
$$

For an even number of observations:

$$
Median = \frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2}
$$

### Numerical Example

Consider:

**10, 20, 30, 40, 50**

There are five observations, and 30 is the middle value.

Therefore:

$$
Median = 30
$$

```python
import statistics

data = [10, 20, 30, 40, 50]

median = statistics.median(data)

print("Median:", median)
```

```python
import statistics

data = [10, 20, 30, 40, 50]

median = statistics.median(data)

print("Median:", median)
```

### Interpretation

The median divides an ordered dataset into two equal parts.

It is less sensitive to extreme values than the mean, making it useful for skewed datasets and datasets containing outliers.

## 3. Mode

### Definition

The mode is the value that occurs most frequently in a dataset.

A dataset can have one mode, multiple modes, or no mode if all values occur with the same frequency.

### Formula

There is no mathematical formula for mode.

The mode is identified by finding the value with the highest frequency.

### Numerical Example

Consider:

**10, 20, 20, 30, 40, 20, 50**

The value 20 occurs three times, which is more frequent than any other value.

Therefore:

**Mode = 20**

```python
from statistics import mode

data = [10, 20, 20, 30, 40, 20, 50]

result = mode(data)

print("Mode:", result)
```

### Business Example

A retail store records the sizes of shoes sold:

**8, 9, 9, 10, 9, 8, 10, 9**

The mode is **9**.

The store can use this information to understand which shoe size is sold most frequently and plan inventory accordingly.

### AI/ML Example

Mode can be used to identify the most frequently occurring category in categorical data.

For example, in a customer dataset, the mode of the "Preferred Payment Method" column can identify the most commonly used payment method.

### Interpretation

The mode represents the most frequently occurring value.

It is particularly useful for categorical data where calculating a mean is not meaningful.

## 4. Weighted Mean

### Definition

A weighted mean is an average in which different values are assigned different levels of importance or weights.

Values with higher weights contribute more to the final average.

### Formula

$$
Weighted\ Mean = \frac{\sum wx}{\sum w}
$$

Where:

- $x$ = value
- $w$ = corresponding weight

### Numerical Example

Suppose a student's grades are:

- Assignment = 80, Weight = 20%
- Midterm = 70, Weight = 30%
- Final Exam = 90, Weight = 50%

Weighted Mean:

$$
(80 \times 0.20) + (70 \times 0.30) + (90 \times 0.50)
$$

$$
= 16 + 21 + 45
$$

$$
= 82
$$

Therefore, the weighted mean is **82**.

```python
scores = [80, 70, 90]
weights = [0.20, 0.30, 0.50]

weighted_mean = sum(
    score * weight for score, weight in zip(scores, weights)
) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

### Business Example

A company evaluates an employee using different performance categories.

For example:

- Technical skills: 40%
- Communication: 20%
- Project delivery: 40%

The weighted mean allows the company to give different importance to each category when calculating an overall score.

### AI/ML Example

Weighted averages are commonly used when combining model outputs, calculating weighted evaluation metrics, or handling observations that have different importance.

For example, a weighted loss function can give greater importance to certain training samples.

### Interpretation

The weighted mean is useful when all values do not have equal importance.

A value with a larger weight has a greater influence on the final result.

## 5. Geometric Mean

### Definition

The geometric mean is an average calculated by multiplying all positive values and taking the nth root, where n is the number of values.

It is particularly useful for growth rates, ratios, percentages, and multiplicative changes.

### Formula

$$
Geometric\ Mean = \sqrt[n]{x_1x_2...x_n}
$$

### Numerical Example

Consider:

**4 and 16**

$$
Geometric\ Mean = \sqrt{4 \times 16}
$$

$$
= \sqrt{64}
$$

$$
= 8
$$

Therefore, the geometric mean is **8**.

```python
import math

data = [4, 16]

geometric_mean = math.prod(data) ** (1 / len(data))

print("Geometric Mean:", geometric_mean)
```

### Business Example

The geometric mean can be used to calculate average investment growth rates over multiple periods because investment returns compound over time.

For example, it can be used when analyzing annual growth rates of a company's revenue.

### AI/ML Example

The geometric mean can be useful when combining multiplicative performance measures or evaluating rates across multiple periods.

For example, it can be used to summarize growth rates or relative performance ratios where multiplication is more appropriate than simple addition.

### Interpretation

The geometric mean is appropriate when values combine multiplicatively.

It is generally used with positive values and is useful for growth rates, ratios, and compounded changes.

## 6. Harmonic Mean

### Definition

The harmonic mean is a type of average that is calculated using the reciprocal of the arithmetic mean of the reciprocals of the values.

It is useful for averaging rates and ratios, especially when the same quantity is involved across different rates.

### Formula

$$
Harmonic\ Mean = \frac{n}{\sum_{i=1}^{n}\frac{1}{x_i}}
$$

### Numerical Example

Consider two values:

**30 and 60**

$$
Harmonic\ Mean = \frac{2}{\frac{1}{30}+\frac{1}{60}}
$$

$$
= 40
$$

Therefore, the harmonic mean is **40**.

```python
data = [30, 60]

harmonic_mean = len(data) / sum(1 / x for x in data)

print("Harmonic Mean:", harmonic_mean)
```

### Business Example

A company may use the harmonic mean when analyzing average rates, such as productivity rates or cost per unit, where the rates need to be combined appropriately.

It is useful when averaging ratios where the denominator remains meaningful.

### AI/ML Example

The harmonic mean is widely used in machine learning to calculate the F1-score.

The F1-score is the harmonic mean of Precision and Recall:

$$
F1 = \frac{2 \times Precision \times Recall}{Precision + Recall}
$$

This makes the F1-score useful when both precision and recall need to be considered together.

### Interpretation

The harmonic mean gives greater influence to smaller values compared with the arithmetic mean.

It is particularly useful for rates, ratios, and metrics such as the F1-score in machine learning.

# Comparison of Measures of Central Tendency

| Measure | Main Purpose | Sensitive to Outliers? | Common Use |
|---|---|---|---|
| Mean | Average value | Yes | Numerical data |
| Median | Middle value | Less sensitive | Skewed data |
| Mode | Most frequent value | Generally no | Categorical data |
| Weighted Mean | Average with different importance | Depends on weights | Scores and business metrics |
| Geometric Mean | Average of multiplicative values | Less suitable for negative values | Growth rates |
| Harmonic Mean | Average of rates/ratios | Strongly influenced by small values | F1-score and rates |

## Summary

The choice of central tendency measure depends on the type and distribution of the data.

- Mean is useful for general numerical data.
- Median is useful when data contains outliers or is skewed.
- Mode is useful for frequently occurring values and categorical data.
- Weighted mean is useful when observations have different importance.
- Geometric mean is useful for growth and multiplicative changes.
- Harmonic mean is useful for rates, ratios, and metrics such as F1-score.
