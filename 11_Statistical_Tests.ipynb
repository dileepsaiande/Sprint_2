## 1. T-Test

### Definition

A t-test is a statistical test used to compare means and determine whether the difference between means is statistically significant.

### Types

- One-sample t-test
- Independent two-sample t-test
- Paired t-test

### Assumptions

- Data is quantitative.
- Observations are independent for an independent-samples test.
- The data is approximately normally distributed, especially for small samples.
- For the standard independent t-test, equal variance is assumed; Welch's t-test can be used when equal variance is not assumed.

### When to Use

Use a t-test when comparing means, especially when the population standard deviation is unknown.

### When Not to Use

Do not use a t-test when its assumptions are seriously violated or when the outcome is categorical.

### Business Example

A company wants to determine whether the average sales before and after a new strategy are different.

### AI/ML Example

A t-test can be used to compare the mean performance of two model configurations under an appropriate experimental design.

### Interpretation

If the p-value is less than the chosen significance level, we reject the null hypothesis and conclude that there is statistical evidence of a difference in means.

```python
from scipy.stats import ttest_ind

group_a = [78, 82, 80, 85, 79, 81, 83]
group_b = [88, 90, 87, 91, 89, 92, 90]

t_stat, p_value = ttest_ind(group_a, group_b)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

## 2. Z-Test

### Definition

A z-test is a statistical test used to compare a sample mean with a population mean, or compare means between groups, when the population standard deviation is known or when large-sample conditions justify a normal approximation.

### Assumptions

- Observations are independent.
- The population standard deviation is known for the classical one-sample z-test.
- The sampling distribution is approximately normal.

### When to Use

Use a z-test when the population standard deviation is known or when the sample size and design justify a normal approximation.

### When Not to Use

Avoid the classical z-test when the population standard deviation is unknown and the sample is small; a t-test is generally more appropriate.

### Business Example

A company knows that the population standard deviation of delivery time is 5 minutes and wants to test whether the average delivery time differs from 30 minutes.

### AI/ML Example

A z-test can be used for certain large-sample comparisons of model metrics or population means when its assumptions are appropriate.

### Interpretation

The z-statistic measures how many standard errors the observed result is away from the null-hypothesis value.

```python
import numpy as np
from scipy.stats import norm

sample = np.array([31, 29, 32, 30, 31, 33, 28, 30, 32, 31])

population_mean = 30
population_std = 5

z_stat = (np.mean(sample) - population_mean) / (
    population_std / np.sqrt(len(sample))
)

p_value = 2 * (1 - norm.cdf(abs(z_stat)))

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

## 3. Chi-Square Test

### Definition

The Chi-Square test is used for categorical data to test relationships between categorical variables or compare observed and expected frequencies.

### Types

1. Chi-Square Test of Independence
2. Chi-Square Goodness-of-Fit Test

### Assumptions

- Data is categorical.
- Observations are independent.
- Expected frequencies should generally be sufficiently large for the chi-square approximation to be reliable.

### When to Use

Use it when analyzing frequency counts for categorical variables.

### When Not to Use

Do not use it directly for comparing means of continuous variables.

### Business Example

A company wants to determine whether customer satisfaction category is associated with customer region.

### AI/ML Example

It can be used to examine relationships between categorical features and a categorical target during exploratory analysis or feature selection.

### Interpretation

A small p-value provides evidence against the null hypothesis of independence, suggesting an association between the categorical variables.

```python
from scipy.stats import chi2_contingency

table = [
    [40, 20],
    [30, 35]
]

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)
```

## 4. ANOVA

### Definition

ANOVA stands for Analysis of Variance.

It is used to test whether the means of three or more groups are statistically different.

### Null Hypothesis

All group means are equal.

### Alternative Hypothesis

At least one group mean is different.

### Assumptions

- Observations are independent.
- The outcome is quantitative.
- The residuals are approximately normally distributed.
- Variances are reasonably similar across groups for classical one-way ANOVA.

### When to Use

Use ANOVA when comparing the means of three or more independent groups.

### When Not to Use

Do not use classical ANOVA when its assumptions are seriously violated or when the outcome is categorical.

### Business Example

A company wants to compare average sales across three different regions.

### AI/ML Example

ANOVA can be used to compare the average performance of several model configurations when the experimental design supports such a comparison.

### Interpretation

If the ANOVA p-value is below the significance level, there is evidence that not all group means are equal.

ANOVA alone does not tell us which specific groups differ. A suitable post-hoc test can be used for that.

```python
from scipy.stats import f_oneway

group_a = [20, 22, 21, 19, 23]
group_b = [25, 27, 26, 24, 28]
group_c = [30, 29, 31, 32, 30]

f_stat, p_value = f_oneway(group_a, group_b, group_c)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

## 5. Mann-Whitney U Test

### Definition

The Mann-Whitney U test is a non-parametric test used to compare two independent groups.

It is based on the ranks of the observations rather than assuming normally distributed data.

### Assumptions

- Two independent groups.
- Observations are independent within and between groups.
- The outcome is at least ordinal.
- For interpreting the test as a difference in distributions, the shapes of the distributions should be considered.

### When to Use

Use it when comparing two independent groups and the assumptions for a t-test are not appropriate, particularly when data is ordinal or strongly non-normal.

### When Not to Use

Do not use it for paired or repeated measurements. For paired data, a Wilcoxon signed-rank test is a different choice.

### Business Example

A company compares customer satisfaction ratings between two independent stores.

### AI/ML Example

It can be used to compare the distributions of a model metric or feature between two independent groups when a non-parametric comparison is appropriate.

### Interpretation

A small p-value provides evidence that the two groups differ in their distributions under the test assumptions.

```python
from scipy.stats import mannwhitneyu

group_a = [70, 72, 68, 75, 71, 69]
group_b = [80, 82, 78, 85, 81, 83]

u_stat, p_value = mannwhitneyu(
    group_a,
    group_b,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)
```

## 6. Shapiro-Wilk Test

### Definition

The Shapiro-Wilk test is used to assess whether a sample is consistent with a normal distribution.

### Hypotheses

H₀: The data is consistent with a normal distribution.

H₁: The data is not consistent with a normal distribution.

### Assumptions

- Observations should be independent.
- The test is designed for assessing normality of quantitative data.

### When to Use

Use it to assess normality, especially when deciding whether a method that assumes normality is reasonable.

### When Not to Use

Do not rely on the test alone for very large datasets because even small deviations from normality can produce a small p-value.

Use graphical methods such as Q-Q plots along with subject-matter knowledge.

### Business Example

A company can check whether a variable is approximately normally distributed before applying a statistical method that assumes normality.

### AI/ML Example

Normality checks can be useful when selecting or validating statistical procedures used during data analysis.

### Interpretation

If the p-value is below the chosen significance level, we reject the null hypothesis and conclude that the data shows evidence against normality.

```python
from scipy.stats import shapiro

data = [10, 12, 11, 13, 12, 14, 15, 11, 13, 12]

stat, p_value = shapiro(data)

print("Shapiro-Wilk Statistic:", stat)
print("P-value:", p_value)
```

# Comparison of Statistical Tests

| Test | Main Purpose | Data Type |
|---|---|---|
| T-Test | Compare means | Numerical |
| Z-Test | Compare means/proportions under normal approximation | Numerical / proportions |
| Chi-Square | Test categorical association or frequency fit | Categorical |
| ANOVA | Compare 3 or more means | Numerical |
| Mann-Whitney U | Compare two independent groups non-parametrically | Ordinal/Numerical |
| Shapiro-Wilk | Test consistency with normality | Numerical |

## Quick Selection Guide

- **Two means → T-Test**
- **Known population standard deviation / appropriate large-sample normal approximation → Z-Test**
- **Categorical variables → Chi-Square**
- **Three or more means → ANOVA**
- **Two independent groups + non-normal/ordinal data → Mann-Whitney U**
- **Check normality → Shapiro-Wilk**
