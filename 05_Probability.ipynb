## 1. Introduction to Probability

### Definition

Probability is a measure of how likely an event is to occur.

The probability of an event ranges from 0 to 1.

- Probability = 0 → Event is impossible
- Probability = 1 → Event is certain
- Probability between 0 and 1 → Event has some chance of occurring

### Formula

For equally likely outcomes:

$$
P(A) = \frac{\text{Number of favorable outcomes}}
{\text{Total number of possible outcomes}}
$$

### Numerical Example

When a fair coin is tossed, there are two possible outcomes:

- Heads
- Tails

The probability of getting Heads is:

$$
P(Heads)=\frac{1}{2}=0.5
$$

Therefore, there is a 50% probability of getting Heads.

```python
probability_heads = 1 / 2

print("Probability of Heads:", probability_heads)
```

### Business Example

A company may estimate the probability that a customer will purchase a product based on historical customer behavior.

For example, if 60 out of 100 similar customers purchased a product, the observed purchase probability is:

$$
P(Purchase)=\frac{60}{100}=0.6
$$

### AI/ML Example

Probability is fundamental to many Machine Learning models.

For example, a classification model may predict:

- Probability of Spam = 0.90
- Probability of Not Spam = 0.10

The model can use these probabilities to make a classification decision.

### Interpretation

Probability helps quantify uncertainty and tells us how likely an event is to occur.

## 4. Independent Events

### Definition

Two events are independent if the occurrence of one event does not affect the probability of the other event.

### Formula

For independent events A and B:

$$
P(A \cap B)=P(A)P(B)
$$

### Numerical Example

Suppose a coin is tossed twice.

The probability of getting Heads on the first toss is:

$$
P(A)=\frac{1}{2}
$$

The probability of getting Heads on the second toss is:

$$
P(B)=\frac{1}{2}
$$

Therefore:

$$
P(A \cap B)=\frac{1}{2}\times\frac{1}{2}
$$

$$
=\frac{1}{4}=0.25
$$

### Python Implementation

```python
p_heads = 0.5

probability_both_heads = p_heads * p_heads

print("Probability of Heads on both tosses:", probability_both_heads)
```

### Business Example

If two unrelated customers independently decide whether to purchase a product, their purchase decisions may be modeled as independent events under an appropriate assumption.

### AI/ML Example

The Naive Bayes algorithm makes a conditional independence assumption between features given the class.

This assumption allows the model to calculate probabilities efficiently.

### Interpretation

Independent events do not influence each other's probabilities.

## 5. Dependent Events

### Definition

Two events are dependent when the occurrence of one event affects the probability of the other event.

### Formula

$$
P(A \cap B)=P(A)P(B|A)
$$

### Numerical Example

Suppose a bag contains:

- 3 red balls
- 2 blue balls

We select two balls without replacement.

Probability of selecting a red ball first:

$$
P(A)=\frac{3}{5}
$$

After selecting one red ball, only 2 red balls remain out of 4 balls.

Therefore:

$$
P(B|A)=\frac{2}{4}
$$

So:

$$
P(A \cap B)=\frac{3}{5}\times\frac{2}{4}
$$

$$
=0.3
$$

```python
probability_first_red = 3 / 5
probability_second_red = 2 / 4

probability_both_red = probability_first_red * probability_second_red

print("Probability of selecting two red balls:", probability_both_red)
```

### Business Example

If a customer has already purchased one product, that information may affect the probability that the same customer purchases another related product.

Therefore, the two events may be dependent.

### AI/ML Example

In real-world datasets, features can be related to each other.

For example, house size and number of bedrooms may not be independent because larger houses may tend to have more bedrooms.

### Interpretation

Dependent events are events where the occurrence of one event changes the probability of another event.

```python
joint_probability = 0.2
probability_b = 0.5

conditional_probability = joint_probability / probability_b

print("P(A|B):", conditional_probability)
```

### Business Example

A company may calculate the probability that a customer purchases a product given that the customer has previously visited the product page.

This can help understand customer behavior.

### AI/ML Example

Conditional probabilities are fundamental to classification models.

For example, a spam classifier may estimate:

$$
P(Spam|Email\ Features)
$$

which represents the probability that an email is spam given its observed features.

### Interpretation

Conditional probability allows us to update the probability of an event based on additional information.

## 7. Bayes' Theorem

### Definition

Bayes' Theorem is used to update the probability of a hypothesis when new evidence becomes available.

### Formula

$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

Where:

- $P(A|B)$ = posterior probability
- $P(B|A)$ = likelihood
- $P(A)$ = prior probability
- $P(B)$ = evidence

### Numerical Example

Suppose:

$$
P(A)=0.1
$$

$$
P(B|A)=0.8
$$

$$
P(B)=0.2
$$

Then:

$$
P(A|B)=\frac{0.8\times0.1}{0.2}
$$

$$
=0.4
$$

Therefore, the posterior probability is **0.4**.

```python
prior = 0.1
likelihood = 0.8
evidence = 0.2

posterior = (likelihood * prior) / evidence

print("Posterior Probability:", posterior)
```

### Business Example

A company can use Bayes' Theorem to update the probability that a customer will churn after observing new information such as reduced usage or complaints.

### AI/ML Example

Bayes' Theorem is the foundation of Bayesian methods and is directly used in algorithms such as Naive Bayes classification.

For example, an email classifier can calculate the probability that an email belongs to the spam class based on the words appearing in the email.

### Interpretation

Bayes' Theorem allows us to update our belief about an event using new evidence.

## 8. Probability Rules

Several important rules are used in probability.

### Rule 1: Complement Rule

$$
P(A^c)=1-P(A)
$$

Example:

If:

$$
P(A)=0.7
$$

then:

$$
P(A^c)=1-0.7=0.3
$$

### Rule 2: Addition Rule

For mutually exclusive events:

$$
P(A\cup B)=P(A)+P(B)
$$

### Rule 3: Multiplication Rule

For independent events:

$$
P(A\cap B)=P(A)P(B)
$$

### Rule 4: Conditional Probability

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

```python
p_a = 0.7

p_not_a = 1 - p_a

print("P(Not A):", p_not_a)
```

### Business Example

Probability rules can help businesses calculate the probability of customers performing different actions, such as purchasing, renewing, or cancelling a service.

### AI/ML Example

Probability rules are used throughout probabilistic machine learning, classification, Bayesian inference, and risk estimation.

### Interpretation

Probability rules provide mathematical methods for combining and updating probabilities.

## 10. Expected Value

### Definition

Expected value is the long-run average value that we expect from a random variable over many repeated trials.

### Formula

For a discrete random variable:

$$
E(X)=\sum x_iP(x_i)
$$

### Numerical Example

Suppose a fair die is rolled.

Possible values:

$$
1,2,3,4,5,6
$$

Each has probability:

$$
\frac{1}{6}
$$

Therefore:

$$
E(X)=
1\left(\frac{1}{6}\right)+
2\left(\frac{1}{6}\right)+
3\left(\frac{1}{6}\right)+
4\left(\frac{1}{6}\right)+
5\left(\frac{1}{6}\right)+
6\left(\frac{1}{6}\right)
$$

$$
E(X)=3.5
$$

Therefore, the expected value is **3.5**.

```python
values = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6

expected_value = sum(
    x * p for x, p in zip(values, probabilities)
)

print("Expected Value:", expected_value)
```

### Business Example

A business can use expected value to estimate the average financial outcome of different decisions.

For example, if a business opportunity has several possible outcomes with known probabilities, expected value can help estimate the long-run average return.

### AI/ML Example

Expected value is used in probabilistic modeling and decision-making.

For example, if a model assigns probabilities to different possible outcomes, expected value can be used to calculate the probability-weighted average outcome.

### Interpretation

Expected value represents the long-run average outcome of a random variable over many repeated trials.

It does not necessarily mean that the expected value will occur in a single trial.
