## 1. Correlation

### Definition

Correlation measures the strength and direction of the relationship between two variables.

The correlation coefficient usually ranges from:

$$
-1 \leq r \leq 1
$$

### Interpretation

- $r = +1$ → Perfect positive correlation
- $r = -1$ → Perfect negative correlation
- $r = 0$ → No linear correlation
- $0 < r < 1$ → Positive correlation
- $-1 < r < 0$ → Negative correlation

### Numerical Example

Suppose:

Study Hours = [1, 2, 3, 4, 5]

Exam Scores = [50, 60, 70, 80, 90]

As study hours increase, exam scores also increase.

Therefore, they have a strong positive correlation.

### Business Example

Advertising expenditure and sales may have a positive relationship.

### AI/ML Example

Correlation can be used during Exploratory Data Analysis to identify relationships between features and the target variable.

### Interpretation

Correlation tells us how strongly two variables are related and in which direction they move together.

Correlation does not by itself prove causation.

```python
import numpy as np

study_hours = [1, 2, 3, 4, 5]
exam_scores = [50, 60, 70, 80, 90]

correlation = np.corrcoef(study_hours, exam_scores)[0, 1]

print("Correlation:", correlation)
```

## 2. Covariance

### Definition

Covariance measures the direction in which two variables change together.

### Formula

Population covariance:

$$
Cov(X,Y)=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{N}
$$

Sample covariance:

$$
Cov(X,Y)=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{n-1}
$$

### Interpretation

- Positive covariance → Variables tend to increase together.
- Negative covariance → One tends to increase when the other decreases.
- Covariance close to zero → Little linear co-movement.

Unlike correlation, covariance is not restricted to the range -1 to +1.

### Numerical Example

Suppose:

X = [1, 2, 3]

Y = [2, 4, 6]

Both variables increase together, so their covariance is positive.

### Business Example

A company may examine how advertising expenditure and sales change together.

### AI/ML Example

Covariance is important in statistical modeling and is used in techniques such as Principal Component Analysis (PCA).

### Interpretation

Covariance tells us the direction of how two variables change together, but its magnitude depends on the units of the variables.

```python
import numpy as np

x = [1, 2, 3]
y = [2, 4, 6]

covariance = np.cov(x, y, ddof=1)[0, 1]

print("Covariance:", covariance)
```

## 3. Positive Correlation

### Definition

Positive correlation occurs when two variables tend to increase or decrease together.

### Example

As study hours increase, exam scores may increase.

### Numerical Example

Study Hours:

[1, 2, 3, 4, 5]

Exam Scores:

[50, 60, 70, 80, 90]

Both variables move in the same direction.

### Business Example

Advertising spending and sales may show positive correlation.

### AI/ML Example

A feature and target variable may have positive correlation, meaning larger feature values tend to be associated with larger target values.

### Interpretation

Positive correlation means the variables tend to move in the same direction.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [50, 60, 70, 80, 90]

plt.scatter(x, y)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Positive Correlation")
plt.show()
```

## 4. Negative Correlation

### Definition

Negative correlation occurs when one variable tends to increase while the other decreases.

### Example

As the price of a product increases, demand may decrease.

### Numerical Example

Price:

[10, 20, 30, 40, 50]

Demand:

[90, 80, 70, 60, 50]

As price increases, demand decreases.

### Business Example

Product price and quantity demanded may have a negative relationship.

### AI/ML Example

During feature analysis, a negative correlation may indicate that larger values of one feature are associated with smaller values of another variable.

### Interpretation

Negative correlation means the variables tend to move in opposite directions.

```python
import matplotlib.pyplot as plt

price = [10, 20, 30, 40, 50]
demand = [90, 80, 70, 60, 50]

plt.scatter(price, demand)
plt.xlabel("Price")
plt.ylabel("Demand")
plt.title("Negative Correlation")
plt.show()
```

## 5. No Correlation

### Definition

No correlation means there is no meaningful linear relationship between two variables.

A correlation coefficient close to zero indicates little or no linear relationship.

### Example

Suppose we compare:

- Employee ID
- Employee salary

The employee ID itself generally should not have a meaningful linear relationship with salary.

### Business Example

A randomly assigned customer ID should not determine customer spending.

### AI/ML Example

During feature selection, a feature with little relationship to another variable may provide limited information about that variable.

However, a correlation close to zero does not prove that there is no relationship of any kind. A nonlinear relationship can still exist.

### Interpretation

A correlation near zero means there is little or no linear relationship between the variables.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

x = np.arange(1, 101)
y = np.random.randint(1, 100, 100)

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Approximately No Linear Correlation")
plt.show()
```

## 6. Pearson Correlation

### Definition

Pearson correlation measures the strength and direction of the linear relationship between two numerical variables.

The Pearson correlation coefficient is represented by:

$$
r
$$

### Formula

$$
r =
\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}
{\sqrt{\sum(X_i-\bar X)^2\sum(Y_i-\bar Y)^2}}
$$

### Range

$$
-1 \leq r \leq 1
$$

### Example

If:

$$
r = 0.90
$$

there is a strong positive linear relationship.

If:

$$
r = -0.90
$$

there is a strong negative linear relationship.

### Business Example

A company can calculate Pearson correlation between advertising expenditure and sales.

### AI/ML Example

Pearson correlation can be used during Exploratory Data Analysis to identify linear relationships among numerical features and the target.

### Important Point

Pearson correlation measures linear association. It does not prove causation.

```python
from scipy.stats import pearsonr

x = [1, 2, 3, 4, 5]
y = [50, 60, 70, 80, 90]

r, p_value = pearsonr(x, y)

print("Pearson Correlation:", r)
print("P-value:", p_value)
```
