## Introduction

A probability distribution describes how the probabilities of different possible outcomes of a random variable are distributed.

Probability distributions are important in Statistics, Data Science, and Machine Learning because they help us model uncertainty and understand the behavior of random variables.

### Distributions Covered

1. Bernoulli Distribution
2. Binomial Distribution
3. Poisson Distribution
4. Uniform Distribution
5. Normal Distribution
6. Exponential Distribution

## 1. Bernoulli Distribution

### Definition

Bernoulli distribution is a discrete probability distribution for an experiment that has only two possible outcomes.

The outcomes are usually represented as:

- Success = 1
- Failure = 0

The experiment is performed only once.

### Formula

The probability mass function is:

$$
P(X=x)=p^x(1-p)^{1-x}
$$

where:

- $x \in \{0,1\}$
- $p$ = probability of success
- $1-p$ = probability of failure

### Numerical Example

Suppose the probability that a customer purchases a product is 0.6.

Therefore:

$$
P(Success)=0.6
$$

$$
P(Failure)=1-0.6=0.4
$$

### Where It Is Used

Bernoulli distribution is used when there are only two possible outcomes.

Examples:

- Yes/No
- Success/Failure
- Purchase/No Purchase
- Spam/Not Spam

### Real-World Example

A customer either purchases a product or does not purchase it.

### AI/ML Example

Binary classification problems have two possible classes, such as spam and not spam.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.6

x = [0, 1]
probabilities = [1 - p, p]

plt.bar(x, probabilities)
plt.xticks([0, 1], ["Failure", "Success"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

### Interpretation

Bernoulli distribution models a single experiment with two possible outcomes.

## 2. Binomial Distribution

### Definition

Binomial distribution models the number of successes in a fixed number of independent Bernoulli trials.

Each trial has:

- Two possible outcomes
- The same probability of success
- Independence between trials

### Formula

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

where:

- $n$ = number of trials
- $k$ = number of successes
- $p$ = probability of success

### Numerical Example

Suppose a customer has a 0.6 probability of purchasing a product.

If we observe 5 customers, we can calculate the probability that exactly 3 customers purchase the product.

Here:

$$
n=5
$$

$$
k=3
$$

$$
p=0.6
$$

### Where It Is Used

Binomial distribution is used when we count successes across a fixed number of independent trials.

### Real-World Example

A company contacts 10 customers and wants to know the probability that exactly 6 customers purchase a product.

### AI/ML Example

Binomial models can be useful for modeling counts of binary outcomes, such as the number of successful predictions in a fixed set of independent trials.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.6

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

### Interpretation

Binomial distribution extends the Bernoulli idea from one trial to multiple independent trials and counts the number of successes.

## 3. Poisson Distribution

### Definition

Poisson distribution is a discrete probability distribution used to model the number of times an event occurs within a fixed interval of time or space.

### Formula

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

where:

- $\lambda$ = average number of events in the interval
- $k$ = number of events
- $e$ = Euler's number

### Numerical Example

Suppose a customer support center receives an average of 4 calls per hour.

Then:

$$
\lambda=4
$$

We can use the Poisson distribution to calculate the probability of receiving exactly 2 calls in one hour.

### Where It Is Used

Poisson distribution is useful for counting events occurring during a fixed interval.

### Real-World Example

Examples include:

- Number of calls received per hour
- Number of website requests per minute
- Number of customers arriving at a store
- Number of system failures per month

### AI/ML Example

Poisson distributions can be used in predictive modeling for count data, such as predicting the number of events occurring within a given period.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 4

x = np.arange(0, 15)
probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

### Interpretation

Poisson distribution models how many times an event occurs within a fixed interval when the average event rate is known.

### Interpretation

Poisson distribution models how many times an event occurs within a fixed interval when the average event rate is known.

## 4. Uniform Distribution

### Definition

A uniform distribution is a probability distribution in which all values within a specified range have equal probability density.

It can be:

- Discrete
- Continuous

### Formula

For a continuous uniform distribution between $a$ and $b$:

$$
f(x)=\frac{1}{b-a}
$$

where:

$$
a \leq x \leq b
$$

### Numerical Example

Suppose a random number is selected between 1 and 10.

Every value within the range has equal probability density.

### Where It Is Used

Uniform distribution is useful when all values within a defined range are equally likely.

### Real-World Example

A simulation may assume that an event can occur at any time within a particular interval with equal likelihood.

### AI/ML Example

Uniform random values can be used in simulations and in some machine learning parameter initialization methods.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

### Interpretation

A uniform distribution has approximately equal probability across its defined range.

## 5. Normal Distribution

### Definition

Normal distribution is a continuous probability distribution with a symmetric bell-shaped curve.

In a perfectly normal distribution:

$$
Mean = Median = Mode
$$

### Formula

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

where:

- $\mu$ = mean
- $\sigma$ = standard deviation

### Numerical Example

Suppose student marks have:

Mean = 70

Standard deviation = 10

Most observations will be concentrated around 70.

### Where It Is Used

Normal distribution is commonly used for modeling continuous measurements that are approximately bell-shaped.

### Real-World Example

Examples may include:

- Measurement errors
- Certain physical measurements
- Standardized test scores

### AI/ML Example

Normal distributions are used in statistical modeling and methods such as Gaussian Naive Bayes and Gaussian mixture models.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(70, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

### Interpretation

Normal distribution is symmetric and bell-shaped. Most observations occur close to the mean, while fewer observations occur farther away.

## 6. Exponential Distribution

### Definition

Exponential distribution is a continuous probability distribution commonly used to model the waiting time until the next event occurs.

### Formula

$$
f(x)=\lambda e^{-\lambda x}
$$

where:

- $\lambda$ = rate parameter
- $x$ = waiting time
- $x \geq 0$

### Numerical Example

Suppose a customer support center receives an average of 2 calls per hour.

Then:

$$
\lambda=2
$$

The exponential distribution can be used to model the waiting time until the next call.

### Where It Is Used

Exponential distribution is commonly used for modeling waiting times between events.

### Real-World Example

Examples include:

- Time until the next customer arrives
- Time between phone calls
- Time between machine failures
- Waiting time between website requests

### AI/ML Example

Exponential distributions can be used in probabilistic models involving event arrival times and survival or reliability analysis.

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_value = 2

data = np.random.exponential(1 / lambda_value, 1000)

plt.hist(data, bins=30)
plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")
plt.show()
```

### Interpretation

The exponential distribution models waiting time between events. It is right-skewed, with many shorter waiting times and fewer longer waiting times.
