# Notebook 10 – Hypothesis Testing

## Introduction

Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population using sample data.

It helps us determine whether there is enough statistical evidence to support a claim about a population.

### Topics Covered

1. Null Hypothesis
2. Alternative Hypothesis
3. p-value
4. Confidence Interval
5. Significance Level
6. Type I Error
7. Type II Error
8. One-tailed Test
9. Two-tailed Test

## 1. Null Hypothesis

### Definition

The null hypothesis, represented by $H_0$, is a statement that there is no effect, no difference, or no relationship between variables.

It is the hypothesis that we test against the evidence in the sample data.

### Example

Suppose a company claims that the average delivery time is 30 minutes.

We can define:

$$
H_0: \mu = 30
$$

This means the population mean delivery time is 30 minutes.

### Business Example

A company introduces a new website and wants to know whether the average purchase amount has changed.

The null hypothesis could be:

$$
H_0: \mu_{new} = \mu_{old}
$$

### AI/ML Example

Suppose we want to compare two model versions.

The null hypothesis could state that there is no difference in their average performance under a specified evaluation procedure.

### Interpretation

The null hypothesis represents the baseline claim that we test using statistical evidence.

## 2. Alternative Hypothesis

### Definition

The alternative hypothesis, represented by $H_1$ or $H_a$, is the statement that there is an effect, difference, or relationship.

It is the hypothesis considered when the null hypothesis is rejected.

### Example

Suppose:

$$
H_0: \mu = 30
$$

Then a two-sided alternative could be:

$$
H_1: \mu \neq 30
$$

This means the average delivery time is different from 30 minutes.

### Types

The alternative hypothesis can be:

#### Two-tailed

$$
H_1: \mu \neq \mu_0
$$

#### Right-tailed

$$
H_1: \mu > \mu_0
$$

#### Left-tailed

$$
H_1: \mu < \mu_0
$$

### Business Example

A company wants to determine whether a new marketing campaign changes average sales.

### AI/ML Example

We can use a hypothesis test to investigate whether two models have different average performance under a defined evaluation procedure.

### Interpretation

The alternative hypothesis represents the effect or difference that the analysis is investigating.

## 3. p-value

### Definition

A p-value is the probability, assuming the null hypothesis is true, of obtaining a test statistic at least as extreme as the one observed.

### Decision Rule

If:

$$
p\text{-value} < \alpha
$$

we reject the null hypothesis.

If:

$$
p\text{-value} \geq \alpha
$$

we fail to reject the null hypothesis.

### Example

Suppose:

$$
p = 0.03
$$

and:

$$
\alpha = 0.05
$$

Since:

$$
0.03 < 0.05
$$

we reject the null hypothesis.

### Important Point

A p-value does NOT tell us the probability that the null hypothesis is true.

### Business Example

A company tests whether a new advertising campaign changes average sales and obtains a p-value of 0.03.

At the 5% significance level, the result would be considered statistically significant under the specified test assumptions.

### AI/ML Example

Statistical tests can be used in model or feature analysis to evaluate specific hypotheses, although model performance should generally be assessed with appropriate validation procedures rather than relying only on p-values.

### Interpretation

A small p-value indicates that the observed data would be relatively unusual if the null hypothesis were true.

```python
from scipy.stats import ttest_1samp

data = [52, 51, 49, 53, 54, 50, 52, 51, 55, 53]

test_statistic, p_value = ttest_1samp(data, popmean=50)

print("Test Statistic:", test_statistic)
print("P-value:", p_value)
```

## 4. Confidence Interval

### Definition

A confidence interval is a range of plausible values for a population parameter based on sample data and a specified confidence level.

### General Form

For a mean under a common large-sample or known-standard-error setting:

$$
\text{Confidence Interval}
=
\text{Estimate}
\pm
\text{Critical Value}\times\text{Standard Error}
$$

### Example

Suppose a sample gives:

Mean = 50

Standard Error = 2

For a 95% confidence interval using a normal critical value of approximately 1.96:

$$
50 \pm 1.96(2)
$$

Therefore:

$$
50 \pm 3.92
$$

Confidence interval:

$$
(46.08,\ 53.92)
$$

### Business Example

A company can estimate the average customer satisfaction score using a confidence interval.

### AI/ML Example

Confidence intervals can be used to quantify uncertainty around estimated model performance metrics or population parameters, when the statistical assumptions and evaluation design support such an interval.

### Interpretation

A 95% confidence interval is constructed by a procedure that, under repeated sampling and its assumptions, captures the true parameter about 95% of the time.

It is not correct to say that there is a 95% probability that a fixed population parameter is inside this particular interval.

```python
import numpy as np
from scipy import stats

data = np.array([48, 50, 51, 49, 52, 50, 53, 47, 51, 49])

confidence_level = 0.95

mean = np.mean(data)
standard_error = stats.sem(data)

confidence_interval = stats.t.interval(
    confidence_level,
    df=len(data) - 1,
    loc=mean,
    scale=standard_error
)

print("Sample Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```
