## 1. Data Cleaning

### Definition
Data cleaning is the process of removing incorrect, duplicate, or inconsistent data before model training.

### How Statistics is Used
Statistics helps identify invalid values using mean, median, standard deviation, and frequency analysis.

### Example
Removing duplicate customer records from a sales dataset.

### Interpretation
Clean data improves model accuracy and reliability.

```python
df = pd.DataFrame({"Age":[25,30,30,np.nan]})
print(df.drop_duplicates())
```

## 2. Missing Value Handling

### Definition
Missing values are empty or null values in a dataset.

### How Statistics is Used
Mean, median, or mode is used to fill missing values.

### Example
Replace missing salary with the average salary.

### Interpretation
Imputation preserves useful information without deleting records.

```python
df = pd.DataFrame({"Salary":[30000,40000,np.nan,50000]})
imputer = SimpleImputer(strategy="mean")
df["Salary"] = imputer.fit_transform(df[["Salary"]])
print(df)
```

## 3. Outlier Detection

### Definition
Outliers are unusually high or low values that differ from the rest of the data.

### How Statistics is Used
IQR and Z-Score methods detect outliers.

### Example
A salary of ₹10,00,000 among salaries around ₹50,000.

### Interpretation
Removing extreme values improves model stability.

```python
salary = pd.Series([30,35,40,45,100])

q1 = salary.quantile(0.25)
q3 = salary.quantile(0.75)
iqr = q3 - q1

outliers = salary[(salary < q1-1.5*iqr) | (salary > q3+1.5*iqr)]
print(outliers)
```

## 4. Feature Engineering

### Definition
Feature engineering creates new useful features from existing data.

### How Statistics is Used
Statistical relationships help generate meaningful features.

### Example
Price per square foot = Price / Area.

### Interpretation
Better features improve prediction quality.

```python
df = pd.DataFrame({
    "Price":[5000000,6000000],
    "Area":[1000,1200]
})

df["Price_per_sqft"] = df["Price"]/df["Area"]
print(df)
```

## 5. Feature Scaling

### Definition
Feature scaling makes numerical features comparable.

### How Statistics is Used
Standardization uses mean and standard deviation.

### Example
Age and Salary have different ranges.

### Interpretation
Scaling improves algorithms like KNN and SVM.

```python
X = pd.DataFrame({"Age":[20,30,40],"Salary":[30000,50000,70000]})

scaler = StandardScaler()
scaled = scaler.fit_transform(X)

print(scaled)
```

## 6. Data Normalization

### Definition
Normalization scales values between 0 and 1.

### How Statistics is Used
Min-Max normalization preserves relative distances.

### Example
Convert salary values into the 0–1 range.

### Interpretation
Useful for distance-based algorithms.

```python
X = pd.DataFrame({"Salary":[30000,50000,70000]})

norm = MinMaxScaler()
print(norm.fit_transform(X))
```

## 7. Model Evaluation

### Definition
Model evaluation measures how well a model performs.

### How Statistics is Used
MAE, RMSE, Accuracy, Precision, and Recall are statistical metrics.

### Example
House price prediction uses MAE.

### Interpretation
Lower error means better predictions.

```python
from sklearn.metrics import mean_absolute_error

y_true=[100,120,140]
y_pred=[105,118,138]

print(mean_absolute_error(y_true,y_pred))
```

## 8. Linear Regression

### Definition
Linear Regression predicts continuous numerical values.

### How Statistics is Used
It models the relationship between input variables and the target using least squares.

### Example
Predicting house prices.

### Interpretation
Higher R² indicates a better fit.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

X = pd.DataFrame({
    "Hours_Studied": [1, 2, 3, 4, 5]
})

y = [35, 45, 55, 65, 75]

model = LinearRegression()
model.fit(X, y)

# Predict using a DataFrame with the same column name
new_data = pd.DataFrame({
    "Hours_Studied": [6]
})

prediction = model.predict(new_data)

print("Predicted Exam Score:", prediction[0])
```

## 9. Logistic Regression

### Definition
Logistic Regression predicts categorical outcomes.

### How Statistics is Used
It estimates probabilities using the logistic function.

### Example
Spam or Not Spam.

### Interpretation
Output is a probability between 0 and 1.

```python
X = [[1],[2],[3],[4]]
y = [0,0,1,1]

model = LogisticRegression()
model.fit(X,y)

print(model.predict([[2.5]]))
```

## 10. Naive Bayes

### Definition
Naive Bayes is a probabilistic classification algorithm.

### How Statistics is Used
It applies Bayes' Theorem and assumes feature independence.

### Example
Email spam detection.

### Interpretation
Fast and effective for text classification.

```python
from sklearn.naive_bayes import GaussianNB

X=[[1],[2],[3],[4]]
y=[0,0,1,1]

model=GaussianNB()
model.fit(X,y)

print(model.predict([[3]]))
```

## 12. Principal Component Analysis (PCA)

### Definition
PCA reduces the number of features while preserving maximum information.

### How Statistics is Used
It uses variance to create principal components.

### Example
Reducing 50 features into 10.

### Interpretation
Lower dimensions improve efficiency.

```python
X=pd.DataFrame({
    "A":[1,2,3],
    "B":[4,5,6]
})

pca=PCA(n_components=1)
print(pca.fit_transform(X))
```

## 13. Recommendation Systems

### Definition
Recommendation systems suggest products or content to users.

### How Statistics is Used

### Example
