## Introduction

Sampling is the process of selecting a smaller group of observations from a larger population for analysis.

Sampling is useful when studying the entire population is expensive, time-consuming, or impractical.

## 1. Sampling

### Definition

Sampling is the process of selecting a subset of individuals or observations from a larger population.

The selected subset is called a sample, while the complete group is called the population.

### Example

Suppose a company has 10,000 customers.

Instead of surveying all 10,000 customers, the company selects 500 customers for a survey.

Population = 10,000 customers

Sample = 500 customers

### Why Sampling Is Used

Sampling can:

- Reduce cost
- Save time
- Make data collection easier
- Allow analysis of large populations

### Business Example

A company can survey a sample of customers to understand overall customer satisfaction.

### AI/ML Example

In Machine Learning, a large dataset may be sampled during exploratory analysis or experimentation to reduce computational requirements.

### Interpretation

A good sample should represent the important characteristics of the population as accurately as possible.

## 2. Random Sampling

### Definition

Random sampling is a sampling method where each member of the population has a known and generally equal chance of being selected.

### Example

Suppose we have 1,000 customers and randomly select 100 customers.

Every customer has an equal opportunity to be selected.

### Python Implementation

```python
import random

population = list(range(1, 101))

sample = random.sample(population, 10)

print("Population:", population)
print("Random Sample:", sample)
```

### Business Example

A company can randomly select 100 customers from its customer database to conduct a satisfaction survey.

### AI/ML Example

Random sampling can be used to create representative subsets of data for experimentation or to split data into training and testing sets.

### Advantages

- Simple to understand
- Reduces selection based on personal choice
- Gives members a fair opportunity to be selected

### Limitation

A random sample may still differ from the population by chance, especially when the sample size is small.

## 3. Stratified Sampling

### Definition

Stratified sampling divides the population into meaningful subgroups called strata and then selects samples from each subgroup.

The strata are usually based on characteristics such as:

- Age
- Gender
- Department
- Region
- Income group

### Example

Suppose a company has:

- 60% customers from Hyderabad
- 40% customers from Bangalore

If we want a sample of 100 customers, we can select:

- 60 from Hyderabad
- 40 from Bangalore

This maintains the population proportions.

### Python Implementation

```python
import pandas as pd

data = pd.DataFrame({
    "City": ["Hyderabad"] * 6 + ["Bangalore"] * 4,
    "Customer_ID": range(1, 11)
})

sample = (
    data.groupby("City", group_keys=False)
        .apply(lambda x: x.sample(frac=0.5, random_state=42))
)

print(sample)
```

### Business Example

A company may divide customers into age groups and select customers from each group so that every age group is represented.

### AI/ML Example

Stratified sampling is commonly used when creating training and testing datasets, especially for classification problems with imbalanced classes.

For example, if a dataset contains 90% non-fraud transactions and 10% fraud transactions, stratified sampling can help preserve this class proportion.

### Interpretation

Stratified sampling helps ensure that important subgroups of the population are represented in the sample.

## 4. Cluster Sampling

### Definition

Cluster sampling divides a population into groups called clusters and randomly selects some clusters for study.

Instead of selecting individuals from every group, the researcher studies members from the selected clusters.

### Example

Suppose a company has offices in 20 cities.

Instead of surveying employees from every city, the company randomly selects 5 cities and surveys employees in those selected cities.

The cities are the clusters.

### Business Example

A retail company may select several stores and collect customer data from those stores.

### AI/ML Example

Cluster sampling can be useful when working with geographically distributed data or very large populations where collecting data from every location is expensive.

### Difference from Stratified Sampling

In stratified sampling:

> We sample from every important subgroup.

In cluster sampling:

> We select some groups and study the selected groups.

### Interpretation

Cluster sampling can reduce the cost and effort of collecting data from geographically or organizationally distributed populations.

## 5. Systematic Sampling

### Definition

Systematic sampling selects observations at a regular interval from an ordered population.

### Formula

The sampling interval is:

$$
k = \frac{N}{n}
$$

where:

- $N$ = population size
- $n$ = desired sample size
- $k$ = sampling interval

### Numerical Example

Suppose:

Population size = 1,000

Desired sample size = 100

Then:

$$
k = \frac{1000}{100}=10
$$

So we can select every 10th observation after choosing a random starting point.

### Python Implementation

```python
population = list(range(1, 101))

k = 10
start = 3

sample = population[start-1::k]

print("Systematic Sample:", sample)
```

### Business Example

A company can select every 10th customer from an ordered customer list for a survey.

### AI/ML Example

Systematic sampling can be used to create a smaller dataset from a large ordered dataset when observations are appropriately arranged.

### Interpretation

Systematic sampling is simple and efficient, but the ordering of the population should be considered because hidden patterns in the ordering can affect the sample.

## 6. Sampling Bias

### Definition

Sampling bias occurs when the method used to select a sample causes the sample to systematically differ from the target population.

### Example

Suppose a company wants to understand customer satisfaction but only surveys customers who gave positive feedback previously.

The sample may not represent all customers.

### Types of Bias

Examples include:

- Selection bias
- Undercoverage
- Voluntary response bias
- Non-response bias

### Business Example

If an online survey is only sent to highly active customers, occasional customers may be underrepresented.

### AI/ML Example

Sampling bias can cause a Machine Learning model to perform well on the sampled data but poorly on the broader population.

For example, if a training dataset does not adequately represent certain customer groups, predictions for those groups may be less reliable.

### How to Reduce Sampling Bias

- Use appropriate sampling methods
- Ensure important groups are represented
- Avoid selecting only convenient observations
- Monitor missing responses
- Compare sample characteristics with the population

### Interpretation

Sampling bias is a systematic problem in the sample selection process and can affect the validity of conclusions.

## 7. Sampling Error

### Definition

Sampling error is the difference between a population parameter and the corresponding statistic calculated from a sample.

It occurs because we study a sample instead of the entire population.

### Numerical Example

Suppose the actual population mean salary is:

₹50,000

A sample of employees produces a mean salary of:

₹48,000

The difference is:

$$
Sampling\ Error = 48,000 - 50,000
$$

$$
= -₹2,000
$$

The magnitude of the sampling error is ₹2,000.

### Important Point

Sampling error can occur even when the sampling method is unbiased because different samples can produce slightly different results.

### Business Example

A customer survey may estimate average satisfaction at 4.2, while the true population average may be 4.3.

The difference is sampling error.

### AI/ML Example

A model trained or evaluated on a sample may produce performance estimates that differ from its performance on the full target population.

### Interpretation

Sampling error is random variation caused by using a sample rather than the entire population.

## 8. Sample Size

### Definition

Sample size is the number of observations included in a sample.

Choosing an appropriate sample size is important because a sample that is too small may produce unstable estimates, while unnecessarily large samples may increase cost and computation.

### Common Factors Affecting Sample Size

Sample size depends on:

- Population size
- Desired confidence level
- Margin of error
- Variability in the population
- Sampling method

### Basic Formula

For a proportion when a large population approximation is appropriate:

$$
n = \frac{Z^2p(1-p)}{E^2}
$$

where:

- $n$ = required sample size
- $Z$ = z-score for the desired confidence level
- $p$ = estimated population proportion
- $E$ = margin of error

### Numerical Example

Suppose:

Confidence level = 95%

$Z = 1.96$

$p = 0.5$

Margin of error = 0.05

Then:

$$
n =
\frac{1.96^2(0.5)(0.5)}
{0.05^2}
$$

Approximately:

$$
n \approx 384
$$

So approximately 384 observations are required under these assumptions for a large population.

### Business Example

A company may calculate the required number of customers to survey when estimating customer satisfaction.

### AI/ML Example

In Machine Learning, sample size affects how well a model can learn patterns and generalize to new data.

More data is not automatically sufficient by itself; the sample should also be relevant and representative.

### Interpretation

An appropriate sample size balances statistical precision, representativeness, cost, and computational resources.
