# Notebook 3 – Measures of Dispersion

## Introduction

Measures of dispersion are statistical measures that describe how spread out or variable the values in a dataset are.

While measures of central tendency such as mean, median, and mode describe the center of the data, measures of dispersion describe how much the observations vary around that center.

### Topics Covered

1. Range
2. Variance
3. Standard Deviation
4. Quartiles
5. Percentiles
6. Interquartile Range (IQR)
7. Coefficient of Variation

## 1. Range

### Definition

Range is the difference between the maximum and minimum values in a dataset.

It is one of the simplest measures of dispersion.

### Formula

$$
Range = Maximum\ Value - Minimum\ Value
$$

### Numerical Example

Consider the dataset:

**10, 20, 30, 40, 50**

Maximum value = 50

Minimum value = 10

$$
Range = 50 - 10
$$

$$
Range = 40
$$

Therefore, the range is **40**.

```python
data = [10, 20, 30, 40, 50]

data_range = max(data) - min(data)

print("Range:", data_range)
```

### Business Example

A company records its monthly sales:

₹50,000, ₹60,000, ₹70,000, ₹80,000, and ₹100,000.

The range is:

₹100,000 - ₹50,000 = ₹50,000.

This tells the company how much the sales values vary between the lowest and highest month.

### AI/ML Example

Range can be used during exploratory data analysis to understand the spread of numerical features.

For example, in a house-price dataset, calculating the range of house prices helps us understand the difference between the cheapest and most expensive properties.

### Interpretation

A larger range indicates a greater difference between the minimum and maximum values.

However, range only uses two observations, so it can be strongly affected by extreme values or outliers.

## 2. Variance

### Definition

Variance measures how far the values in a dataset are spread from the mean.

A higher variance indicates that the observations are more spread out, while a lower variance indicates that the observations are closer to the mean.

### Population Variance Formula

$$
\sigma^2 = \frac{\sum (x_i-\mu)^2}{N}
$$

Where:

- $x_i$ = individual observation
- $\mu$ = population mean
- $N$ = number of observations

### Sample Variance Formula

$$
s^2 = \frac{\sum (x_i-\bar{x})^2}{n-1}
$$

### Numerical Example

Consider:

**2, 4, 6**

Mean:

$$
Mean = \frac{2+4+6}{3}=4
$$

Squared differences:

$$
(2-4)^2 = 4
$$

$$
(4-4)^2 = 0
$$

$$
(6-4)^2 = 4
$$

Population variance:

$$
Variance = \frac{4+0+4}{3}
$$

$$
Variance = \frac{8}{3} \approx 2.67
$$

```python
import statistics

data = [2, 4, 6]

population_variance = statistics.pvariance(data)
sample_variance = statistics.variance(data)

print("Population Variance:", population_variance)
print("Sample Variance:", sample_variance)
```

### Business Example

A company can use variance to understand how much monthly sales fluctuate.

If the variance of sales is high, sales may be changing significantly from month to month.

### AI/ML Example

Variance is important in Machine Learning because it helps us understand the spread of numerical features.

It is also related to feature scaling and dimensionality reduction techniques such as Principal Component Analysis (PCA), where variance is used to understand how much information is represented by different directions in the data.

### Interpretation

Variance measures the average squared distance of observations from the mean.

Because the differences are squared, variance is expressed in squared units. Therefore, standard deviation is often easier to interpret.

## 3. Standard Deviation

### Definition

Standard deviation measures how much the values in a dataset typically differ from the mean.

It is the square root of variance.

### Formula

$$
\sigma = \sqrt{\frac{\sum(x_i-\mu)^2}{N}}
$$

For a sample:

$$
s = \sqrt{\frac{\sum(x_i-\bar{x})^2}{n-1}}
$$

### Numerical Example

Consider:

**2, 4, 6**

Mean = 4

Population variance:

$$
\frac{(2-4)^2+(4-4)^2+(6-4)^2}{3}
=
\frac{8}{3}
$$

Therefore:

$$
Standard\ Deviation = \sqrt{\frac{8}{3}}
$$

$$
\approx 1.63
$$

```python
import statistics

data = [2, 4, 6]

std_dev = statistics.pstdev(data)

print("Population Standard Deviation:", std_dev)
```

### Business Example

A company can use standard deviation to understand the consistency of its monthly sales.

A low standard deviation means sales are relatively consistent, while a high standard deviation indicates greater variation.

### AI/ML Example

Standard deviation is commonly used in Machine Learning for standardization.

Standardization transforms a feature so that it has a mean close to 0 and a standard deviation close to 1.

The common formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

This can help algorithms that are sensitive to feature scales.

### Interpretation

A low standard deviation means observations tend to be close to the mean.

A high standard deviation means observations are more spread out from the mean.

Standard deviation is easier to interpret than variance because it has the same units as the original data.

## 4. Quartiles

### Definition

Quartiles divide an ordered dataset into four equal parts.

The three main quartiles are:

- Q1 – First Quartile
- Q2 – Second Quartile
- Q3 – Third Quartile

Q2 is the median.

### Meaning

- **Q1:** Approximately 25% of observations are below Q1.
- **Q2:** Approximately 50% of observations are below Q2.
- **Q3:** Approximately 75% of observations are below Q3.

### Numerical Example

Consider:

**10, 20, 30, 40, 50, 60, 70**

The median (Q2) is:

**40**

The first quartile (Q1) is approximately:

**20**

The third quartile (Q3) is approximately:

**60**

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

### Business Example

A company can divide employee salaries into quartiles to understand salary distribution.

For example, Q1 represents the lower part of the salary distribution, Q2 represents the median, and Q3 represents the upper part.

### AI/ML Example

Quartiles are useful during exploratory data analysis and can help identify the distribution of features and potential outliers.

### Interpretation

Quartiles help us understand how observations are distributed across different parts of a dataset.

## 5. Percentiles

### Definition

A percentile indicates the value below which a certain percentage of observations fall.

For example, the 25th percentile means that approximately 25% of the observations are below that value.

### Common Percentiles

- 25th percentile
- 50th percentile
- 75th percentile
- 90th percentile
- 95th percentile
- 99th percentile

### Numerical Example

Consider:

**10, 20, 30, 40, 50, 60, 70**

The 50th percentile represents the median.

The 25th and 75th percentiles help describe the lower and upper portions of the dataset.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

for p in [25, 50, 75, 90]:
    print(f"{p}th Percentile:", np.percentile(data, p))
```

### Business Example

A company can use the 90th percentile of customer waiting time to understand a service-level threshold.

For example, if the 90th percentile waiting time is 15 minutes, approximately 90% of observed waiting times are at or below that value.

### AI/ML Example

Percentiles can be used to understand feature distributions and identify extreme observations.

For example, values above the 99th percentile may be investigated as potential outliers.

### Interpretation

Percentiles help us understand the relative position of observations within a dataset.

They are particularly useful for analyzing distributions, performance thresholds, and outliers.

## 6. Interquartile Range (IQR)

### Definition

The Interquartile Range (IQR) measures the spread of the middle 50% of a dataset.

It is the difference between the third quartile and the first quartile.

### Formula

$$
IQR = Q3 - Q1
$$

### Numerical Example

Suppose:

Q1 = 20

Q3 = 60

Then:

$$
IQR = 60 - 20
$$

$$
IQR = 40
$$

Therefore, the IQR is **40**.

import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

### Business Example

A company can use IQR to analyze the middle 50% of employee salaries while reducing the influence of extremely high or low salaries.

### AI/ML Example

IQR is commonly used for detecting potential outliers during data preprocessing.

A common rule is:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds can be investigated as potential outliers.

### Interpretation

A larger IQR means the middle 50% of observations are more spread out.

IQR is more resistant to extreme values than the range because it focuses on the middle portion of the data.

## 7. Coefficient of Variation

### Definition

The Coefficient of Variation (CV) measures the relative amount of variability in a dataset compared with its mean.

It is expressed as a percentage.

### Formula

$$
CV = \frac{Standard\ Deviation}{Mean} \times 100
$$

### Numerical Example

Suppose:

Mean = 50

Standard Deviation = 5

Then:

$$
CV = \frac{5}{50} \times 100
$$

$$
CV = 10\%
$$

Therefore, the coefficient of variation is **10%**.

```python
mean = 50
standard_deviation = 5

cv = (standard_deviation / mean) * 100

print("Coefficient of Variation:", cv, "%")
```

### Business Example

A business can use the coefficient of variation to compare the relative variability of two products or investments, even when their average values are different.

For example, if two investments have different average returns, CV can help compare their variability relative to their average.

### AI/ML Example

Coefficient of variation can be used during exploratory data analysis to compare the relative variability of numerical features that have different scales or means.

It can help identify features with relatively high variability.

### Interpretation

A lower coefficient of variation indicates lower variability relative to the mean.

A higher coefficient of variation indicates greater variability relative to the mean.

CV is useful when comparing variability across datasets with different scales.
