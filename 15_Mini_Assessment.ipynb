### Question 1

Write Python code to calculate the Mean, Median, and Variance.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
```

### Question 2

Detect Outliers using the IQR Method.

```python
import pandas as pd

salary = pd.Series([30, 35, 40, 45, 50, 120])

Q1 = salary.quantile(0.25)
Q3 = salary.quantile(0.75)

IQR = Q3 - Q1

outliers = salary[(salary < Q1 - 1.5*IQR) | (salary > Q3 + 1.5*IQR)]

print("Outliers:")
print(outliers)
```

### Question 3

Calculate the Correlation Matrix and display a Heatmap.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Age":[22,25,30,35,40],
    "Salary":[25000,30000,40000,50000,65000],
    "Experience":[1,2,4,6,8]
})

corr = df.corr()

plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap="YlGnBu")
plt.title("Correlation Heatmap")
plt.show()
```

### Question 4

Perform Feature Scaling using StandardScaler.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

X = pd.DataFrame({
    "Age":[20,30,40],
    "Salary":[30000,50000,70000]
})

scaler = StandardScaler()

scaled = scaler.fit_transform(X)

print(scaled)
```

### Question 5

Normalize the data using MinMaxScaler.

### Question 6

Train a Linear Regression model and predict one value.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

X = pd.DataFrame({
    "Hours_Studied":[1,2,3,4,5]
})

y = [35,45,55,65,75]

model = LinearRegression()

model.fit(X,y)

new_data = pd.DataFrame({"Hours_Studied":[6]})

prediction = model.predict(new_data)

print("Predicted Score:", prediction[0])
```

### Question 7

Apply K-Means Clustering with 2 clusters.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
from sklearn.cluster import KMeans

data = pd.DataFrame({
    "Height": [150,155,160,170,175,180],
    "Weight": [45,50,52,70,75,80]
})

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
data["Cluster"] = kmeans.fit_predict(data)

print(data)
```

### Question 8

Calculate the VIF of the given features.

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = pd.DataFrame({
    "Age":[25,30,35,40,45],
    "Income":[30,42,49,63,68],
    "Salary":[35,45,55,60,72]
})

vif = pd.DataFrame()

vif["Feature"] = X.columns

vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print(vif)
```

### Question 9

Perform a Shapiro-Wilk Test.

```python
from scipy.stats import shapiro

data = [10,12,11,13,12,14,15,11,13,12]

stat, p = shapiro(data)

print("Statistic:", stat)
print("P-value:", p)
```

### Question 10

Perform a T-Test on two sample groups.

```python
from scipy.stats import ttest_ind

group1 = [78,80,82,81,79]

group2 = [88,90,91,89,87]

t_stat, p_value = ttest_ind(group1, group2)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
